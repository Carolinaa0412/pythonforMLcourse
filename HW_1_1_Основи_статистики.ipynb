**Завдання 3.** Для того ж набору даних в змінній `sum_purchases` обчисліть **медіану** кожним з зазначених методів:  

  3.1. без використання жодних бібліотек, користуючись тільки Python.  
  3.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран. Порівняйте результати зі значенням вибіркового середнього. Напишіть коментар, для яких задач ви б використовували вибіркове середнє, а для яких - медіану?

```python
sum_purchases = [316, 493, 367, 153, 224, 121, 322, 450, 196, 397,
    281, 239, 484, 124, 223, 448, 432, 255, 4310, 3524]

# пайтон
def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    if n % 2 == 1:
        return sorted_data[n // 2]
    else:
        mid1 = sorted_data[n // 2 - 1]
        mid2 = sorted_data[n // 2]
        return (mid1 + mid2) / 2

median_pure_python = calculate_median(sum_purchases)

# бібліотека NumPy
median_numpy = np.median(sum_purchases)

print(f"Медіана на Python: {median_pure_python}")
print(f"Медіана з NumPy: {median_numpy}")
```

**Завдання 4**. Обчислість стандартне відхилення наведеної вибірки в змінній `sum_purchases` кожним з зазначених методів:  

  4.1. без використання жодних бібліотек, користуючись тільки Python.  
  4.2. використовуючи відповідний метод бібліотеки numpy.  

Виведіть результат обчислень в кожному з випадків на екран.

```python
sum_purchases = [316, 493, 367, 153, 224, 121, 322, 450, 196, 397,
    281, 239, 484, 124, 223, 448, 432, 255, 4310, 3524]

mean_val = sum(sum_purchases) / len(sum_purchases)
variance = sum((x - mean_val) ** 2 for x in sum_purchases) / (len(sum_purchases) - 1)
std_pure_python = variance ** 0.5

std_numpy = np.std(sum_purchases, ddof=1).round(2)

print(f"Стандартне відхилення на Python: {std_pure_python}")
print(f"Стандартне відхилення з NumPy: {std_numpy}")
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
sum_purchases = [ 316,  493,  367,  153,  224,  121,  322,  450,  196,  397,  281,  239,  484,  124,  223,  448,  432,  255, 4310, 3524]
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,  7,  4,  2]

def pearson_corr(x, y):
    n = len(x)
    mean_x, mean_y = sum(x) / n, sum(y) / n
    numerator = sum((x[i] - mean_x) * (y[i] - mean_y) for i in range(n))
    denom_x = sum((x[i] - mean_x) ** 2 for i in range(n)) ** 0.5
    denom_y = sum((y[i] - mean_y) ** 2 for i in range(n)) ** 0.5
    return numerator / (denom_x * denom_y)
corr_pure_python = pearson_corr(sum_purchases, visits)

corr_matrix = np.corrcoef(sum_purchases, visits)
corr_numpy = corr_matrix[0, 1]

print(f"Коефіцієнт Пірсона на Python: {corr_pure_python:.4f}")
print(f"Коефіцієнт Пірсона з NumPy: {corr_numpy:.4f}")
```

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

Я дійшла до такого висновку: більшість точок (наші 18 покупців з 20) зібрані зліва внизу, вони роблять покупки в межах від 100 до 500 грн. Але праворуч чітко видно дві відірвані точки - це аномально великі покупки (викиди) на 3524 та 4310 грн.

**Чи варто щось змінити? Так, звісно.** Коефіцієнт кореляції Пірсона дуже чутливий до викидів, тому ці два великі рахунки сильно спотворюють картину і штучно тягнуть показник зв'язку.

Перед підрахунком коефіцієнта Пірсона можливо варто видалити ці викиди, адже вони не є типовими для більшості клієнтів.

**Завдання 7**. В попередньому завданні ми помітили, що викиди нам заважають отримати "правдиве" значення кореляції. Давайте вилучимо записи, які відповідають значенням з викидами в сумі покупок і обчисліть коеф. кореляції. Для цього поставте `threshold`, який дорінює середньому значенню + 2 стандартних вдіхилля суми покупок і всі значення, які більше за нього, будемо вважати викидами (є і інші правила для визначення викидів - з ним и познайомимось пізніше).

Вилучіть значення з обох масивів і обчисліть коеф. кореляції Пірсона з допомогою `numpy` для кількості візитів і суми покупок без записів-викидів. Виведіть результат на екран, побудуйте діаграму розсіювання і проаналізуйте результат: що можемо сказати про залежність в даних?

```python
import matplotlib.pyplot as plt

sum_purchases = np.array([316, 493, 367, 153, 224, 121, 322, 450, 196, 397, 281, 239, 484, 124, 223, 448, 432, 255, 4310, 3524])
visits = np.array([6, 2, 10, 3, 9, 9, 2, 3, 6, 6, 10, 2, 5, 10, 3, 5, 4, 7, 4, 2])

# поріг для визначення викидів
mean_p = np.mean(sum_purchases)
std_p = np.std(sum_purchases)
threshold = mean_p + 2 * std_p

# прибираємо викиди з обох масивів
mask = sum_purchases <= threshold
sum_purchases_clean = sum_purchases[mask]
visits_clean = visits[mask]

corr_matrix = np.corrcoef(sum_purchases_clean, visits_clean)
corr_clean = corr_matrix[0, 1]

print(f"Порогове значення: {threshold:.2f}")
print(f"Коефіцієнт кореляції Пірсона без викидів: {corr_clean:.4f}")

plt.figure(figsize=(8, 5))
plt.scatter(sum_purchases_clean, visits_clean, color='blue', alpha=0.7)
plt.title('Графік рахунків vs. Клієнтів (без викидів)')
plt.xlabel('Сума рахунку (UAH)')
plt.ylabel('Кількість клієнтів')
plt.grid(True)
plt.show()
```
